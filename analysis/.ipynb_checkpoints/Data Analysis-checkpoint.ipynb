```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("./korean_skincare_master.csv")
```

```python
# 2. Fix chemical numbers (e.g. 1,2-Hexanediol) so commas inside ingredient names don't break splitting
corrected_ingredients = df["Ingredients"].dropna().str.replace(r"\b1,\s*2-", "1.2-", regex=True)

# 3. Split, clean, and count individual ingredients
all_ingredients = (
    corrected_ingredients
    .str.split(",")
    .explode()
    .str.strip()
    .str.title()
)

print("=== Top 20 Most Common Ingredients ===")
print(all_ingredients.value_counts().head(20))

# 4. Tag Hero Actives accurately
actives = {
    "Hyaluronic Acid": "hyaluronic acid|sodium hyaluronate|hydrolyzed hyaluronic acid",
    "Panthenol": "panthenol",
    "Niacinamide": "niacinamide",
    "Adenosine": "adenosine",
    "Centella Asiatica": "centella",
    "Ceramides": "ceramide",
    "Tea Tree Oil": "melaleuca alternifolia|tea tree",
    "Salicylic Acid (BHA)": "salicylic acid",
    "Retinol / Retinal": "retinol|retinal",
    "Snail Mucin": "snail secretion filtrate"
}

active_counts = {name: df["Ingredients"].str.contains(kw, case=False, na=False).sum() for name, kw in actives.items()}
active_df = pd.DataFrame(list(active_counts.items()), columns=["Active", "Count"]).sort_values(by="Count", ascending=False)

print("\n=== Hero Active Product Counts ===")
print(active_df.to_string(index=False))
```

# Part 1: Pricing & Promotional Strategy
- Which brands position themselves in the premium tier vs. budget-friendly everyday routine tier (evaluated via median prices)?

```python
# Group by brand to calculate median price and product count
brand_pricing = (
    df.groupby("Brand")
    .agg(Product_Count=("Product Name", "count"), Median_Price=("Current Price", "median"))
    .reset_index()
)
```

```python
# Filter for brands with 5+ products to ensure statistical relevance
valid_brands = brand_pricing[brand_pricing["Product_Count"] >= 5]
```

```python
# Top Premium Tier (Highest Median Price)
premium_tier = valid_brands.sort_values(by="Median_Price", ascending=False).head(10)
```

```python
# Top Budget-Friendly Tier (Lowest Median Price)
budget_tier = valid_brands.sort_values(by="Median_Price", ascending=True).head(10)
```

```python
print("--- Premium Tier ---")
print(premium_tier.to_string(index=False))

print("\n--- Budget-Friendly Tier ---")
print(budget_tier.to_string(index=False))
```

- Which product categories (e.g., Sunscreens, Cleansers, Serums) carry the deepest average and maximum discounts?

```python
# Calculate average and maximum discounts by category
discount_analysis = (
    df.groupby("Category_Standardized")
    .agg(
        Product_Count=("Product Name", "count"),
        Avg_Discount=("Discount (%)", "mean"),
        Max_Discount=("Discount (%)", "max"),
    )
    .reset_index()
    .sort_values(by="Avg_Discount", ascending=False)
)

print(discount_analysis.to_string(index=False))
```

- Is there a measurable correlation between product price and customer review ratings, or do budget formulas perform equally well?


```python
# 1. Compute correlation matrix between Price, Rating, and Review Count
correlation_matrix = df[["Current Price", "Rating", "Review_Count"]].corr()
print("--- Correlation Matrix ---")
print(correlation_matrix)

# 2. Segment by Price Tiers (Budget, Mid, Premium) and analyze ratings & reviews
df["Price_Tier"] = pd.qcut(
    df["Current Price"], q=3, labels=["Budget", "Mid", "Premium"]
)
tier_performance = (
    df.groupby("Price_Tier")
    .agg(
        Product_Count=("Product Name", "count"),
        Avg_Rating=("Rating", "mean"),
        Median_Rating=("Rating", "median"),
        Avg_Reviews=("Review_Count", "mean"),
    )
    .reset_index()
)

print("\n--- Performance by Price Tier ---")
print(tier_performance.to_string(index=False))
```

- Price has virtually no relationship with customer review ratings. Expensive formulations are not rated higher than affordable ones.
- Higher Engagement in the Budget Tier: Budget-tier products average 114.7 reviews, compared to 53.8 for mid-tier and 28.0 for premium-tier products, showing that everyday staples attract significantly higher customer volume and feedback.
- Median ratings remain remarkably steady across all three tiers (4.56 for budget, 4.52 for mid, and 4.58 for premium), demonstrating that affordable K-beauty formulations perform just as exceptionally as high-end alternatives.

```python
# Group by brand to calculate total review volume
top_review_brands = (
    df.groupby("Brand")["Review_Count"]
    .sum()
    .reset_index(name="Total_Reviews")
    .sort_values(by="Total_Reviews", ascending=False)
    .head(10)
)

print("--- Top 10 Brands by Review Volume ---")
print(top_review_brands.to_string(index=False))
```

- What is the rating distribution across categories: are certain routine steps evaluated consistently higher than others?


```python
# Group by category to analyze rating distribution across routine steps
rating_by_category = (
    df.groupby("Category_Standardized")
    .agg(
        Product_Count=("Product Name", "count"),
        Mean_Rating=("Rating", "mean"),
        Median_Rating=("Rating", "median"),
        Std_Rating=("Rating", "std"),
    )
    .reset_index()
    .sort_values(by="Median_Rating", ascending=False)
)

print(rating_by_category.to_string(index=False))
```

- Essence, Moisturizer, Toner, Mask, Cleanser, and Serum all cluster tightly at the top with median ratings ranging between 4.55 and 4.63, demonstrating consistently high baseline satisfaction across essential daily skincare steps.
- Specialized treatments (like Eye Care, Lip Care, Exfoliators, Treatments, and Sunscreens) maintain strong median ratings ($\sim4.38$ to $4.50$), with Lip Care showing the lowest standard deviation ($0.85$), indicating very uniform, reliable customer agreement on performance.

```python
# Identify all hero ingredient indicator columns
ingredient_cols = [col for col in df.columns if col.startswith("Has_")]

print("=== PART 1: Top Catalog Items by Hero Ingredient ===")
# Example: Find top-rated products containing Centella Asiatica with high review counts
centella_favorites = (
    df[(df["Has_Centella"] == 1) & (df["Rating"].notna())]
    .sort_values(by="Rating", ascending=False)[
        ["Brand", "Product Name", "Rating", "Current Price"]
    ]
    .head(10)
)
print("\nTop Centella Asiatica Products:")
print(centella_favorites.to_string(index=False))


print("\n=== PART 2: Correlation & Rating Impact of Hero Ingredients ===")
# Evaluate how active formulations correlate with review ratings compared to baseline
ingredient_performance = []

for col in ingredient_cols:
    ingredient_name = col.replace("Has_", "").replace("_", " ")

    # Segment products with vs. without the ingredient
    with_active = df[df[col] == 1]["Rating"]
    without_active = df[df[col] == 0]["Rating"]

    ingredient_performance.append(
        {
            "Hero_Ingredient": ingredient_name,
            "Product_Count": len(with_active),
            "Avg_Rating_With": round(with_active.mean(), 2),
            "Avg_Rating_Without": round(without_active.mean(), 2),
            "Rating_Difference": round(
                with_active.mean() - without_active.mean(), 3
            ),
        }
    )

# Convert summary to DataFrame and sort by impact
impact_df = pd.DataFrame(ingredient_performance).sort_values(
    by="Avg_Rating_With", ascending=False
)
print(impact_df.to_string(index=False))
```

```python
price_comparison = []

for col in ingredient_cols:
    ingredient_name = col.replace("Has_", "").replace("_", " ")
    
    # Segment prices for products containing the active vs. without
    price_with = df[df[col] == 1]["Current Price"]
    price_without = df[df[col] == 0]["Current Price"]
    
    price_comparison.append({
        "Hero_Ingredient": ingredient_name,
        "Product_Count": len(price_with),
        "Median_Price_With": round(price_with.median(), 2),
        "Mean_Price_With": round(price_with.mean(), 2),
        "Median_Price_Without": round(price_without.median(), 2)
    })

# Convert to DataFrame and sort by median price
price_df = pd.DataFrame(price_comparison).sort_values(by="Median_Price_With", ascending=False)

print("=== Hero Ingredient Price Positioning (Median vs Mean) ===")
print(price_df.to_string(index=False))
```

```python
# 1. Define keyword patterns for potentially sensitizing or irritating ingredients
# Searching directly through the full 'Ingredients' text column
irritant_keywords = {
    "Drying_Alcohol": ["alcohol denat", "sd alcohol", "ethanol", "isopropyl alcohol"],
    "Synthetic_Fragrance": ["parfum", "fragrance", "artificial fragrance"],
    "Essential_Oils": [
        "lavender oil",
        "tea tree oil",
        "citrus oil",
        "bergamot oil",
        "eucalyptus oil",
        "peppermint oil",
    ],
    "Harsh_Surfactants": [
        "sodium lauryl sulfate",
        "sls",
        "sodium laureth sulfate",
        "sles",
    ],
    "Potential_Irritant_Witch_Hazel": ["witch hazel", "hamamelis"],
}

# 2. Create indicator flags by scanning the raw ingredient strings
for category, keywords in irritant_keywords.items():
  # Check if any keyword matches the product's ingredient list (case-insensitive)
  pattern = "|".join(keywords)
  df[f"Contains_{category}"] = (
      df["Ingredients"].astype(str).str.lower().str.contains(pattern, na=False)
  )

# 3. Analyze prevalence and impact on average ratings
analysis_results = []
for category in irritant_keywords.keys():
  col_name = f"Contains_{category}"
  with_item = df[df[col_name] == True]["Rating"]
  without_item = df[df[col_name] == False]["Rating"]

  analysis_results.append({
      "Category": category,
      "Products_Containing": len(with_item),
      "Percentage_of_Catalog": round(
          (len(with_item) / len(df)) * 100, 1
      ),
      "Avg_Rating_With": round(with_item.mean(), 2) if len(with_item) > 0 else 0,
      "Avg_Rating_Without": round(without_item.mean(), 2),
  })

summary_df = pd.DataFrame(analysis_results)
print("--- Ingredient Safety & Irritant Text Analysis ---")
print(summary_df.to_string(index=False))

# 4. Filter clean products for sensitive skin (e.g., free from drying alcohol, synthetic fragrance, and harsh sulfates)
sensitive_skin_safe = df[
    (df["Contains_Drying_Alcohol"] == False)
    & (df["Contains_Synthetic_Fragrance"] == False)
    & (df["Contains_Harsh_Surfactants"] == False)
].copy()

print(
    f"\nTotal products safe from drying alcohols, synthetic fragrances, and"
    f" sulfates: {len(sensitive_skin_safe)}"
)
```

```python
# 1. Direct boolean filtering using your existing Contains_ columns
def _get_safe_subset(subset_df):
  """Excludes products where any irritant flag is True."""
  irritant_cols = [
      "Contains_Drying_Alcohol",
      "Contains_Synthetic_Fragrance",
      "Contains_Essential_Oils",
      "Contains_Harsh_Surfactants",
      "Contains_Potential_Irritant_Witch_Hazel",
  ]

  # Only filter against columns that actually exist in the DataFrame
  active_irritant_cols = [
      col for col in irritant_cols if col in subset_df.columns
  ]

  if not active_irritant_cols:
    return subset_df

  # Keep rows where ALL irritant columns are False (or 0)
  # (df[cols] == False).all(axis=1) handles both bool and 0/1 integer flags cleanly
  is_safe = (subset_df[active_irritant_cols] == False).all(axis=1)
  safe_subset = subset_df[is_safe].copy()
  return safe_subset
```

```python
def _calculate_composite_score(subset):
  """Helper function to compute multi-criteria score."""
  qualified = subset.copy()

  # Normalize metrics between 0 and 1
  qualified["Rating_Score"] = qualified["Rating"] / 5.0

  max_log_reviews = np.log1p(subset["Review_Count"].max())
  qualified["Popularity_Score"] = (
      np.log1p(qualified["Review_Count"]) / max_log_reviews
  )

  max_price = qualified["Current Price"].max()
  min_price = qualified["Current Price"].min()
  if max_price > min_price:
    qualified["Value_Score"] = 1 - (
        (qualified["Current Price"] - min_price) / (max_price - min_price)
    )
  else:
    qualified["Value_Score"] = 1.0

  # Composite Score (50% Rating, 30% Review Volume, 20% Price Value)
  qualified["Composite_Score"] = (
      0.50 * qualified["Rating_Score"]
      + 0.30 * qualified["Popularity_Score"]
      + 0.20 * qualified["Value_Score"]
  )
  return qualified
```

```python
def find_best_product_for_ingredient(
    df, ingredient_col, exclude_irritants=True, min_reviews=None):
  # 1. Verify ingredient column exists
  if ingredient_col not in df.columns:
    return f"Ingredient column '{ingredient_col}' not found in dataset."

  # 2. Isolate products containing the specific active ingredient
  subset = df[df[ingredient_col] == 1].copy()

  # 3. Filter out irritants using existing Contains_ columns
  if exclude_irritants:
    subset = _get_safe_subset(subset)

  if len(subset) == 0:
    return f"No safe products found containing {ingredient_col}."

  # 4. Apply review volume threshold
  if min_reviews is None:
    min_reviews = (
        subset["Review_Count"].median()
        if "Review_Count" in subset.columns
        else 10
    )

  qualified = subset[subset["Review_Count"] >= min_reviews].copy()
  if len(qualified) == 0:
    qualified = subset.copy()

  # 5. Normalize metrics and compute Composite Score
  qualified["Rating_Score"] = qualified["Rating"] / 5.0

  max_log_reviews = np.log1p(subset["Review_Count"].max())
  qualified["Popularity_Score"] = (
      np.log1p(qualified["Review_Count"]) / max_log_reviews
      if max_log_reviews > 0
      else 1.0
  )

  max_price = qualified["Current Price"].max()
  min_price = qualified["Current Price"].min()
  if max_price > min_price:
    qualified["Value_Score"] = 1 - (
        (qualified["Current Price"] - min_price) / (max_price - min_price)
    )
  else:
    qualified["Value_Score"] = 1.0

  qualified["Composite_Score"] = (
      0.50 * qualified["Rating_Score"]
      + 0.30 * qualified["Popularity_Score"]
      + 0.20 * qualified["Value_Score"]
  )

  # 6. Return top-ranked product
  best_match = qualified.sort_values(
      by="Composite_Score", ascending=False
  ).iloc[0]
  solution = best_match[
      ["Brand", "Product Name", "Current Price", "Rating", "Review_Count"]
  ]
  return solution
```

```python
# 1. Function dedicated exclusively to Skin Type filtering
def find_best_product_for_skin_type(
    df, skin_type_col, exclude_irritants=True, min_reviews=None):
  if skin_type_col not in df.columns:
    return f"Skin type column '{skin_type_col}' not found in dataset."

  subset = df[df[skin_type_col] == 1].copy()

  if exclude_irritants:
    subset = _get_safe_subset(subset)

  if len(subset) == 0:
    return f"No safe products found for skin type: {skin_type_col}"

  if min_reviews is None:
    min_reviews = subset["Review_Count"].median()

  qualified = subset[subset["Review_Count"] >= min_reviews]
  if len(qualified) == 0:
    qualified = subset

  scored = _calculate_composite_score(qualified)
  best_match = scored.sort_values(by="Composite_Score", ascending=False).iloc[0]
  return best_match[
      ["Brand", "Product Name", "Current Price", "Rating", "Review_Count"]]
```
